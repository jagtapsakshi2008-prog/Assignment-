```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
import joblib

# Load dataset (use your dataset path)
df = pd.read_csv("employee_attrition.csv")

# Convert categorical to numeric (example)
df['JobSatisfaction'] = df['JobSatisfaction'].map({'Low':1, 'Medium':2, 'High':3})
df['OverTime'] = df['OverTime'].map({'No':0, 'Yes':1})

# Features and target
X = df[['Age','MonthlyIncome','DistanceFromHome','YearsAtCompany','JobSatisfaction','OverTime']]
y = df['Attrition'].map({'No':0, 'Yes':1})

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Train model
model = RandomForestClassifier()
model.fit(X_train, y_train)

# Save model
joblib.dump(model, "employee_attrition_model.pkl")

print("✅ Model saved successfully!")
```
