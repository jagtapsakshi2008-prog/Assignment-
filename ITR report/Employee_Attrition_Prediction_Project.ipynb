# Data Preprocessing

```python
# Drop Unnecessary Columns

df.drop(["EmployeeCount",
         "StandardHours",
         "Over18",
         "EmployeeNumber"],
        axis=1,
        inplace=True)
```

```python
# Display Dataset

df.head()
```

```python
"""
The unnecessary columns have been removed because they do not
contribute to employee attrition prediction.
""";
```

```python
# Encode Attrition Column

df["Attrition"] = df["Attrition"].map({"Yes":1,
                                       "No":0})
```

```python
# Display Target Values

df["Attrition"].value_counts()
```

```python
"""
The target variable has been converted into numerical values.

Yes = 1

No = 0
""";
```

```python
# Import Label Encoder

from sklearn.preprocessing import LabelEncoder
```

```python
# Create Label Encoder Object

le = LabelEncoder()
```

```python
# Encode Gender

df["Gender"] = le.fit_transform(df["Gender"])
```

```python
# Encode OverTime

df["OverTime"] = le.fit_transform(df["OverTime"])
```

```python
"""
Binary categorical columns are converted into numeric values
using Label Encoding.
""";
```

```python
# One Hot Encode Remaining Categorical Columns

df = pd.get_dummies(df,
                    drop_first=True)
```

```python
# Display Dataset

df.head()
```

```python
"""
Multi-category columns have been converted into dummy variables
using One Hot Encoding.
""";
```

```python
# Separate Input and Output

X = df.drop("Attrition", axis=1)

y = df["Attrition"]
```

```python
# Display Shape

print(X.shape)

print(y.shape)
```

```python
"""
X contains all input features.

y contains the target variable.
""";
```

```python
# Import Train Test Split

from sklearn.model_selection import train_test_split
```

```python
# Split Dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
"""
The dataset has been divided into:

80% Training Data

20% Testing Data

Stratified sampling is used to maintain the same class
distribution in both datasets.
""";
```

```python
# Import Standard Scaler

from sklearn.preprocessing import StandardScaler
```

```python
# Create StandardScaler Object

scaler = StandardScaler()
```

```python
# Scale Training Data

X_train = scaler.fit_transform(X_train)
```

```python
# Scale Testing Data

X_test = scaler.transform(X_test)
```

```python
"""
Feature scaling standardizes the numerical values.

It helps machine learning algorithms perform better by
bringing all features to a similar scale.
""";
```

```python
# Check Class Distribution

y_train.value_counts()
```

```python
"""
The training data is moderately imbalanced.

While the project allows using SMOTE, it also specifies that
using class_weight='balanced' in the models is an acceptable
approach. We'll use class_weight='balanced' in the classifiers
to address the imbalance.
""";
```

# Models to Build

# Logistic Regression

```python
# Import Logistic Regression

from sklearn.linear_model import LogisticRegression
```

```python
# Create Logistic Regression Model

lr = LogisticRegression(
    class_weight="balanced",
    random_state=42
)
```

```python
# Train Logistic Regression Model

lr.fit(X_train, y_train)
```

```python
# Predict Testing Data

y_pred_lr = lr.predict(X_test)
```

```python
# Predict Probability

y_prob_lr = lr.predict_proba(X_test)[:,1]
```

```python
# Import Evaluation Metrics

from sklearn.metrics import accuracy_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_auc_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
# Accuracy

accuracy_score(y_test, y_pred_lr)
```

```python
# Recall

recall_score(y_test, y_pred_lr)
```

```python
# Recall

recall_score(y_test, y_pred_lr)
```

```python
# Confusion Matrix

cm = confusion_matrix(y_test, y_pred_lr)

cm
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(y_test, y_pred_lr);
```

```python
"""
The confusion matrix shows:

• True Positive
• True Negative
• False Positive
• False Negative

It helps us understand how well the model correctly predicts employee attrition.
""";
```

```python
# Classification Report

print(classification_report(y_test, y_pred_lr))
```

```python
"""
The classification report displays:

• Precision
• Recall
• F1-Score
• Support

Recall is one of the most important metrics for this project because identifying employees who are likely to leave is the primary business goal.
""";
```

# Decision Tree

```python
# Import Decision Tree

from sklearn.tree import DecisionTreeClassifier
```

```python
# Create Decision Tree Model

dt = DecisionTreeClassifier(
    class_weight="balanced",
    max_depth=5,
    random_state=42
)
```

```python
# Train Decision Tree

dt.fit(X_train, y_train)
```

```python
# Predict Testing Data

y_pred_dt = dt.predict(X_test)

y_prob_dt = dt.predict_proba(X_test)[:,1]
```

```python
# Accuracy

accuracy_score(y_test, y_pred_dt)
```

```python
# Recall

recall_score(y_test, y_pred_dt)
```

```python
# Confusion Matrix

cm_dt = confusion_matrix(y_test, y_pred_dt)

cm_dt
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_dt);
```

```python
# Classification Report

print(classification_report(y_test, y_pred_dt))
```

# Random Forest

```python
# Import Random Forest

from sklearn.ensemble import RandomForestClassifier
```

```python
# Create Random Forest Model

rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    class_weight="balanced",
    random_state=42
)
```

```python
# Train Random Forest Model

rf.fit(X_train, y_train)
```

```python
# Predict Test Data

y_pred_rf = rf.predict(X_test)
```

```python
# Predict Probability

y_prob_rf = rf.predict_proba(X_test)[:,1]
```

```python
# Accuracy

accuracy_score(y_test, y_pred_rf)
```

```python
# Recall

recall_score(y_test, y_pred_rf)
```

```python
# F1 Score

f1_score(y_test, y_pred_rf)
```

```python
# Confusion Matrix

cm_rf = confusion_matrix(y_test, y_pred_rf)

cm_rf
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_rf);
```

```python
"""
The Random Forest model generally provides better prediction
performance by combining multiple decision trees.

It reduces overfitting and improves overall accuracy.
""";
```

# Model Comparison

```python
# Compare Models

result = pd.DataFrame({

    "Model":[
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],

    "Accuracy":[
        accuracy_score(y_test,y_pred_lr),
        accuracy_score(y_test,y_pred_dt),
        accuracy_score(y_test,y_pred_rf)
    ],

    "Recall":[
        recall_score(y_test,y_pred_lr),
        recall_score(y_test,y_pred_dt),
        recall_score(y_test,y_pred_rf)
    ],

    "F1 Score":[
        f1_score(y_test,y_pred_lr),
        f1_score(y_test,y_pred_dt),
        f1_score(y_test,y_pred_rf)
    ]

})

result
```

```python
# Display Best Model

result.sort_values(by="Recall",
                   ascending=False)
```
