```python
import seaborn as sns
from sklearn.preprocessing import LabelEncoder

taxis = sns.load_dataset("taxis").drop(columns=["pickup", "dropoff"])


def label_encode_dataframe(df):
    df = df.copy()
    encoders = {}

    for col in df.select_dtypes(include=["object", "category"]).columns:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col].astype(str))
        encoders[col] = le

    return df, encoders


encoded_taxis, encoders = label_encode_dataframe(taxis)

print(encoded_taxis.head())
```
