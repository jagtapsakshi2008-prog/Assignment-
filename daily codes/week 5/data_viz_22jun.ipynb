```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([10,23,29,34,39,45,50])
y=np.array([14,21,31,43,48,55,58])
z=np.array([56,50,44,37,26,20,16])
           
```

### line chart

```python
# it is used to show the changes of the data over the time period (contineous data)
plt.plot(x,y)
```

```python
plt.xlabel(" x value")
plt.ylabel(" y value")
plt.title( " line chart : x vs y")
plt.plot(x,y ,marker='o',mfc='red',ls='--')

# marker parameters: o,*,H,<,>,^
```

```python
plt.xlabel(" X value")
plt.ylabel(" Z label")
plt.title(" line chart : x vs z")
plt.plot(x,y,marker='o',mfc='red',ls='--')
```

```python
plt.plot(x,z,marker='*',mfc='green')
plt.xlabel(" X value")
plt.ylabel(" Z label")
plt.title(" line chart : x vs y")
plt.plot(x,y,marker='o',mfc='red',ls='--',label='2024')
plt.plot(x,y,marker='o',mfc='black',ls='--',color='orange',label='2025')
plt.grid()
plt.legend()
```

```python
plt.figure(figsize=(16,9))
plt.subplot(2,2,1)
plt.xlabel(" X value")
plt.ylabel(" Y label")
plt.title(" line chart : x vs y")
plt.plot(x,y,marker='o',mfc='red',ls='--')
plt.grid()

plt.subplot(2,2,2)
plt.xlabel(" X value")
plt.ylabel(" Y label")
plt.title(" line chart : x vs y")
plt.plot(x,y,marker='o',mfc='red',ls='--')
plt.grid()
```

# Bar chart
