```python
# we need to be specific about what kind of error we want to catch
```

```python
try:
    f=open("a1.txt")
    print(var1)
except FileNotFoundError:
    print("file does not exist")
```

```python
try:
    f=open("a1.txt")
    print(var1)
except FileNotFoundError:
    print("file does not exist")
except:
    print("something went wrong")
```

```python
try:
    f=open("a1.txt")
    print(var1)
except FileNotFoundError:
    print("file does not exist")
except Exception as e:
    print(e)
```

```python
# else => if try block doesn't raise any exception then else block will run 
```

```python
try:
    f=open("a1.txt")
except FileNotFoundError:
    print("file does not exist")
except Exception as e:
    print(e)
else:
    print(f.read())
    f.close()

    
```

```python
try:
    f=open("a1.txt")
    print(var1)
except FileNotFoundError:
    print("file does not exist")
except Exception as e:
    print(e)
else:
    print(f.read())
    f.close()
finally:
    print("run complete")
```

```python
# manually raise exception
```

```python
try:
    f=open("a1.txt")
    if f.name=='a1.txt':
        raise Exception
except FileNotFoundError:
    print("file does not exist")
except Exception as e:
    print(e)
else:
    print(f.read())
    f.close()
finally:
    print("run complete")
    
```

```python
# take 2 numbers as user input
# write program for division of these 2 numbers
# handle ZeroDivisionError and ValueError 
```

```python
try:
    num1=int(input("enter the number 1:"))
    num2=int(input("enter the number 2:"))
    res=num1/num2
    print("division=>",res)

except ZeroDivisionError:
    print("division by zero cannot be allowed")

except ValueError:
    print("enter valid value")
```
