```python
# 1. write a function safe_divide(a,b) that divides two numbers and handles ZeroDivisionError and 
# TypeError seperately , printing an appropriate message for each . use a finally block to print 
# "operation attempted . ' regardless of the outcome . test it with valid numbers , division by zero , and 
# a string input .
```

```python
def safe_divide(a,b):
    try:
        result=a/b
        return result
    except ZeroDivisionError:
        return "Error: Division by zero is not allowed"
    except TypeError:
        return "Error: Both input must be number"      
```

```python
print(safe_divide(30,4))
print(safe_divide(7,0))
print(safe_divide(10,'2'))
```

```python
# 2. Write a function validate_age(age) that raises a custom exception InvalidAgeError (a class you define, inheriting from Exception)
# if age is negative or greater than 120.
# Catch this exception in a calling block and print a user-friendly message.
# Test it with at least 3 different values.
```

```python
class InvalidAgeError(Exception):
    pass
    
def validate_age(age):
    if age < 0 or age > 120:        
        raise InvalidAgeError("age must be betn 0 to 120")
    print(f"valid age:{age}")

test=[25,-5,122]
for age in test:
    try:
        validate_age(age)
    except InvalidAgeError as e:
        print(f"Invalid Age Error:{e}")
```

```python
# 3. Write a program that repeatedly asks the user (use a loop with at least 3 simulated inputs in a list,
# e.g., user_inputs = ['10', 'abc', '0', '5']) to enter a number and divide 100 by it.

# Handle the following in a single try block using multiple except clauses:
# a) ValueError if the input is not a number.
# b) ZeroDivisionError if the input is zero.
# c) A generic Exception as a catch-all for anything else.
# Print the result if successful; otherwise print which error occurred.
```

```python
user_input=['10','abc','0','5']
for val in user_input:
    try:
        num=int(val)
        res=100/num
        print(f"100/{num}={res}")
    except ValueError:
        print(f"ValueError: '{val}'mis not a valid number")
    except ZeroDivisionError:
        print(f"ZeroDivisionError: cannott divide ny zero")
    except Exception as e:
        print(f"Error:{e}")
```

```python
import numpy as np
a=np.arange(1,21)
print("array=>",a)

b=a.reshape(4,5) # array reshaping into 4*5 2D array
print("\n 4*5 2D reshaped array=>", b)

print("\n sum of each row:")
print(np.sum(b,axis=1))

print("\nSum of each column:")
print(np.sum(b,axis=0))

element=a[a>10]
print("\n element greater than 10:")
print(element) # element greater than 10 

a[a%2==0]=-1
print("\n array after replacing even numbers with -1")
print(a)

```

```python
# 7. Create a Pandas DataFrame from the dictionary below and perform the following:
# data = {
#    "Name": ["Aarav", "Priya", "Karan", "Meera", "Sahil"],
#    "Age": [23, 21, 25, 22, 24],
#    "Department": ["AI", "Data Science", "AI", "Web Dev", "Data Science"],
#   "Score": [88, 92, 79, 95, 67]
#  }

# a) Display the first 3 rows and the data types of each column.
# b) Add a new column Grade based on Score:
# Score ≥ 90 → 'A'
# Score ≥ 75 → 'B'
# Else → 'C'

# c) Sort the DataFrame by Score in descending order.
# d) Filter rows where Department is 'AI'.
```

```python
import pandas as pd

data = {
    "Name": ["Aarav", "Priya", "Karan", "Meera", "Sahil"],
    "Age": [23, 21, 25, 22, 24],
    "Department": ["AI", "Data Science", "AI", "Web Dev", "Data Science"],
    "Score": [88, 92, 79, 95, 67]
}

df = pd.DataFrame(data)

print("First 3 Rows:")
print(df.head(3))
print("\nData Types:")
print(df.dtypes)

def grade(score):
    if score >= 90:
        return 'A'
    elif score >= 75:
        return 'B'
    else:
        return 'C'

df["Grade"] = df["Score"].apply(grade)
print("\nDataFrame with Grade:")
print(df)

sorted_df = df.sort_values(by="Score", ascending=False)
print("\nSorted by Score (Descending):")
print(sorted_df)

ai_df = df[df["Department"] == "AI"]
print("\nEmployees in AI Department:")
print(ai_df)
```

```python
# 8. Using the same DataFrame from Q7
# a) Group by Department and find the average Score for each.
# b) Find the row with the highest Score using idxmax().
# c) Rename the column Score to Marks.
# d) Drop the Age column and display the resulting DataFrame.
```

```python

print(df.groupby("Department")["Score"].mean()) # Average Score by Department
print("\n")

print(df.loc[df["Score"].idxmax()]) # Row with highest Score
print("\n")

df.rename(columns={"Score": "Marks"}, inplace=True) # Rename Score to Marks
print("\n")
print(df)

df.drop("Age", axis=1, inplace=True) # Drop Age column
print("\n")
print(df)
```

```python
# 9. Create a DataFrame with some missing values as shown below and handle them:
# import numpy as np
# data = {
#   "Product": ["Pen", "Notebook", "Eraser", "Sharpener", "Ruler"],
#   "Price": [10, np.nan, 5, 8, np.nan],
#   "Stock": [100, 50, np.nan, 30, 20]
# }

# a) Check the number of missing values in each column using isnull().sum().
# b) Fill missing Price values with the column mean.
# c) Drop rows where Stock is still missing.
# d) Save the cleaned DataFrame to a CSV file named cleaned_products.csv without the index.
```

```python
import pandas as pd
import numpy as np

data = {
    "Product": ["Pen", "Notebook", "Eraser", "Sharpener", "Ruler"],
    "Price": [10, np.nan, 5, 8, np.nan],
    "Stock": [100, 50, np.nan, 30, 20]
}

df = pd.DataFrame(data)

print(df.isnull().sum())# Count missing values

df['Price']=df["Price"].fillna(df["Price"].mean())#  Fill missing Price values with mean

df.dropna(subset=["Stock"], inplace=True)#  Drop rows where Stock is missing

df.to_csv("cleaned_products.csv", index=False)#  Save to CSV

print(df)
```
